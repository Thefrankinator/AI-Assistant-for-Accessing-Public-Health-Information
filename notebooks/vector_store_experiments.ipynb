```python
from langchain_community.vectorstores import FAISS # type: ignore
from langchain_openai import OpenAIEmbeddings  # type: ignore
```

```python
librairy = FAISS.load_local("C:/School/Autres/NLP/AAFAPHI/AI-Assistant-for-Accessing-Public-Health-Information/data/faiss_vector_store",  
    OpenAIEmbeddings(model="text-embedding-3-large",dimensions=1024),
    allow_dangerous_deserialization=True)
```

```python
text = " Quelle vacin doit faire un enfant entre 5-6 ans ?"
docs = librairy.similarity_search(query = text, k=2)
for doc in docs:
    print(doc.page_content)
```
